```python
import graphviz

# ==========================================
# 1. SITEMAP DIAGRAM
# ==========================================
sitemap = graphviz.Digraph("Sitemap", format="png")
sitemap.attr(
    bgcolor="#121216", rankdir="TB", nodesep="0.3", ranksep="0.6"
)

# Default style for nodes and edges
sitemap.attr(
    "node",
    fontname="Arial",
    style="filled,rounded",
    shape="box",
    fontcolor="#ffffff",
    height="0.4",
    width="1.1",
    penwidth="0",
)
sitemap.attr("edge", color="#666666", penwidth="1.5", arrowsize="0.6")

# Level 1 - Root Node
sitemap.node("Home", "Home", fillcolor="#e63946", fontcolor="#ffffff")

# Level 2 - Main Navigation Nodes
sitemap.node("Events", "Events", fillcolor="#06d6a0", fontcolor="#000000")
sitemap.node("Register", "Register", fillcolor="#ef476f", fontcolor="#ffffff")
sitemap.node("Schedule", "Schedule", fillcolor="#ffd166", fontcolor="#000000")
sitemap.node("Contact", "Contact", fillcolor="#118ab2", fontcolor="#ffffff")

# Level 3 - Subpages
subpages = {
    "Events": ["Seminars", "Concerts", "Sports"],
    "Register": ["Select Pass", "Details Form", "Confirmation"],
    "Schedule": ["Timeline", "Venue Map"],
    "Contact": ["FAQ", "Support"],
}

for parent, children in subpages.items():
    for child in children:
        node_id = f"{parent}_{child}"
        sitemap.node(node_id, child, fillcolor="#1d3557", fontcolor="#ffffff")
        sitemap.edge(parent, node_id)

# Connect Root to Level 2
for main_nav in ["Events", "Register", "Schedule", "Contact"]:
    sitemap.edge("Home", main_nav)

# Save and render
sitemap.render("sitemap_output", cleanup=True)
print("Sitemap saved as sitemap_output.png")


# ==========================================
# 2. FLOWCHART DIAGRAM
# ==========================================
flowchart = graphviz.Digraph("Flowchart", format="png")
flowchart.attr(
    bgcolor="#121216", rankdir="TB", nodesep="0.4", ranksep="0.5"
)

flowchart.attr("edge", color="#ffffff", fontcolor="#ffffff", penwidth="1.2", arrowsize="0.7")

# Node Definitions
flowchart.node("Start", "Start", shape="ellipse", style="filled", fillcolor="#2b2d42", fontcolor="#ffffff")
flowchart.node("Browse", "Browse Events", shape="box", style="filled,rounded", fillcolor="#1d3557", fontcolor="#ffffff")
flowchart.node("Select", "Select Event", shape="box", style="filled,rounded", fillcolor="#1d3557", fontcolor="#ffffff")
flowchart.node("Check", "Tickets\nAvailable?", shape="diamond", style="filled", fillcolor="#333333", fontcolor="#ffffff")
flowchart.node("SoldOut", "Show 'Sold Out'", shape="box", style="filled,rounded", fillcolor="#1d3557", fontcolor="#ffffff")
flowchart.node("Fill", "Fill Form", shape="box", style="filled,rounded", fillcolor="#1d3557", fontcolor="#ffffff")
flowchart.node("GetPass", "Get Pass", shape="box", style="filled,rounded", fillcolor="#1d3557", fontcolor="#ffffff")
flowchart.node("End", "End", shape="ellipse", style="filled", fillcolor="#2b2d42", fontcolor="#ffffff")

# Flow Connections
flowchart.edge("Start", "Browse")
flowchart.edge("Browse", "Select")
flowchart.edge("Select", "Check")
flowchart.edge("Check", "SoldOut", label=" No ")
flowchart.edge("Check", "Fill", label=" Yes ")
flowchart.edge("Fill", "GetPass")
flowchart.edge("SoldOut", "End")
flowchart.edge("GetPass", "End")

# Save and render
flowchart.render("flowchart_output", cleanup=True)
print("Flowchart saved as flowchart_output.png")
```
